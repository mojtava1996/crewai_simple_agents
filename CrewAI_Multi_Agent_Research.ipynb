# CrewAI Multi-Agent Research Team — Google Colab

A simple beginner-friendly CrewAI project.

**Agents**
- 🔎 Researcher: gathers and organizes information
- ✍️ Writer: turns the research into a concise report

**Goal:** Learn Agents → Tasks → Crew → kickoff.

## 2. Imports

````python
import os
from getpass import getpass
from crewai import Agent, Task, Crew, Process, LLM

print("CrewAI imported successfully.")
````

## 3. Configure the LLM

This example uses an OpenAI-compatible API. Enter your API key when prompted.
Do not hard-code secrets into a notebook that you upload to GitHub.


````python
API_KEY = getpass("Enter your API key: ")

MODEL = "gpt-4o-mini"
BASE_URL = "https://api.openai.com/v1"

os.environ["OPENAI_API_KEY"] = API_KEY

llm = LLM(
    model=MODEL,
    api_key=API_KEY,
    base_url=BASE_URL,
    temperature=0.2,
)

print("LLM configured.")
````

## 4. Create the Researcher Agent

````python
researcher = Agent(
    role="Technology Researcher",
    goal="Find useful and relevant information about a technology topic.",
    backstory=(
        "You are a careful technology researcher. "
        "You organize information clearly and focus on important facts."
    ),
    llm=llm,
    verbose=True,
)
````

## 5. Create the Writer Agent

````python
writer = Agent(
    role="Technical Writer",
    goal="Turn research into a clear, useful report for a beginner.",
    backstory=(
        "You are an experienced technical writer who explains complex "
        "AI concepts using simple language and practical examples."
    ),
    llm=llm,
    verbose=True,
)
````

## 6. Create Tasks

````python
topic = "CrewAI and multi-agent AI systems"

research_task = Task(
    description=f'''
Research the following topic:

{topic}

Explain:
1. What it is
2. How it works
3. Three practical use cases
4. Main advantages
5. Main limitations

Keep the research factual and organized.
''',
    expected_output="A structured research brief with the requested sections.",
    agent=researcher,
)

writing_task = Task(
    description='''
Using the research produced by the researcher, write a beginner-friendly
report about the topic.

Include:
- A short introduction
- How it works
- Three practical examples
- Advantages
- Limitations
- A short conclusion

Keep it concise and easy to understand.
''',
    expected_output="A polished 500-700 word beginner-friendly report.",
    agent=writer,
)
````

## 7. Create the Crew

````python
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, writing_task],
    process=Process.sequential,
    verbose=True,
)

print("Crew created.")
````

## 8. Run the Crew

````python
result = crew.kickoff()

print("\n" + "=" * 80)
print("FINAL REPORT")
print("=" * 80)
print(result)
````

````python
result=crew.kickoff_async()

print("\n" + "=" * 80)
print("FINAL REPORT")
print("=" * 80)
print(result)
````

## 9. Understand the workflow

```text
User topic
   ↓
Researcher Agent
   ↓
Research Task
   ↓
Writer Agent
   ↓
Writing Task
   ↓
Final Report
```

**Agent** = AI worker  
**Task** = work assigned to an agent  
**Crew** = team of agents and tasks  
**Process.sequential** = tasks run in sequence  
**kickoff()** = starts the workflow
